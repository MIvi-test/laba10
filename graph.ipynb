```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,5))
```

```python
f = open('data.txt').readlines()
data = {}
for i in f:
    key,value = list(map(int,i.split(' ')))
    data[key] = value

```

```python
ax = sns.lineplot(data)
ax.grid(True)
ax.set(title = 'Зависимость размера файла от размера блока', xlabel = 'Разбер блока, байты', ylabel = "размер закодированного файла, байты")
```
